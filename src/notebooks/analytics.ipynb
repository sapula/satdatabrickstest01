```python
%sql
WITH max_date_cte AS (
  SELECT MAX(o_orderdate) AS max_date
  FROM tpccopy001.poc.orders
)
SELECT
  l.l_partkey AS product_id,
  p.p_name,
  p.p_mfgr,
  p.p_brand,
  p.p_type,
  p.p_retailprice,
  SUM(l.l_quantity) AS total_sold
FROM tpccopy001.poc.orders o
JOIN tpccopy001.poc.lineitem l ON o.o_orderkey = l.l_orderkey
JOIN tpccopy001.poc.part p ON l.l_partkey = p.p_partkey
CROSS JOIN max_date_cte
WHERE o.o_orderdate > DATE_SUB(max_date_cte.max_date, 30)
GROUP BY l.l_partkey, p.p_name, p.p_mfgr, p.p_brand, p.p_type, p.p_retailprice
ORDER BY total_sold DESC
LIMIT 1
```
